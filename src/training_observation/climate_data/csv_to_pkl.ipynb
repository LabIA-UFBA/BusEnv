```python
import pandas as pd
import pickle

df1 = pd.read_csv("INMET_NE_BA_A401_SALVADOR_01-01-2024_A_31-12-2024.CSV", sep = ';', encoding = 'latin1', skiprows=8) 
df2 = pd.read_csv("INMET_NE_BA_A401_SALVADOR_01-01-2025_A_30-09-2025.CSV", sep = ';', encoding = 'latin1', skiprows=8) 

df_total = pd.concat([df1, df2], ignore_index=True)
df_total
```

```python
import pandas as pd

# Renomenar columnes brasileres → angleses
df = df_total.rename(columns={
    "Data": "date",
    "Hora UTC": "hour",
    "PRECIPITAÇÃO TOTAL, HORÁRIO (mm)": "precip"
})

# Convertir data i precip
df["date"] = pd.to_datetime(df["date"], format="%Y/%m/%d")
df["precip"] = df["precip"].astype(str).str.replace(",", ".").astype(float)

# Seleccionar només les columnes útils
df_clean = df[["date", "hour", "precip"]]

# Tallar a partir de l’1 de març de 2024
df_clean = df_clean[df_clean["date"] >= "2024-03-01"]
df_clean
```

```python
with open("climate_data_2.pkl", "wb") as f:
    pickle.dump(df, f) 
```
